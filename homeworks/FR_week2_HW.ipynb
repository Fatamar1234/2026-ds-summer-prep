## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def no_dup(inputs):
    empty_set = set()
    for i in inputs:
        if i in empty_set:
            return False
        
        empty_set.add(i)
    return True

numbers = [1,2, 3, 4, 5]
print(no_dup(numbers))
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python

# Find the key with the highest value


import pandas as pd
# for reasing files 
with open("lil-wayne.txt", "r", encoding="utf-8") as file:
    new_diction = {}
    lines = file.read().upper()

    for character in lines:
    # if the character is already in the dictionary, increment its count
        if character in new_diction:
            new_diction[character] += 1
        # else just one 
        else:
            new_diction[character] = 1
most_common_character = max(new_diction, key=new_diction.get)
print(" the character that appears the most is: '  ", most_common_character, "  ' and it appears ", new_diction[most_common_character], " times")


#now we will sort the dictionary using the sorted fucntion 
sort_var = sorted(new_diction.items(), key=lambda x: x[1])
print(sort_var)
````
